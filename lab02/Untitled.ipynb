```python
from playwright.async_api import async_playwright

from IPython.display import display, Image



BASE_URL = "https://www.cthulhuclub.com/"



async with async_playwright() as p:

  # 在课程 Docker 中启动无头浏览器

  browser = await p.chromium.launch(headless=True)

  try:

    # Context 隔离会话数据，Page 表示这个环境中的一个标签页

    context = await browser.new_context()

    page = await context.new_page()

    # 访问页面；动态图书列表可能在页面加载后继续请求数据

    await page.goto(BASE_URL)



    print(await page.title())

    print(page.url)



    # 截图并在 Notebook 中显示

    await page.screenshot(path="miniread.png", full_page=True)

    display(Image(filename="miniread.png"))

  finally:

    # 关闭浏览器及其中的页面；退出 async with 后停止 Playwright

    await browser.close()
```
